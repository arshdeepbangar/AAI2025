# Exercise 3 - Self-Reflection Prompt for Improving Output

**Goal:** Ask the AI to critique an original summary against clear requirements and then produce an improved version.

**Tools used:** ChatGPT  and Google Colab

## Source text to summarize

Prompt engineering is the practice of designing clear, detailed instructions for an AI system so it can produce a useful result. Different techniques include role-playing prompts, content structuring, negative prompting, prompt chaining, and back-and-forth refinement. Prompt chaining links multiple prompts so the output from one step becomes input for the next, while refinement improves a response through repeated feedback.

## Original summary - Before self-reflection

> Prompt engineering is about giving AI better prompts. There are different methods, including chaining and refinement, that can improve the response.

## Self-reflection prompt

You are a careful editor. Critique the original summary using the requirements below, then write a revised summary.

1. **Accuracy:** Keep the important ideas from the source and do not add new facts.
2. **Clarity:** Use simple language that a high school student could understand.
3. **Audience:** Write for a beginner MIS student.
4. **Length:** Exactly 2 sentences and 35-50 words total.
5. **Focus:** Define prompt engineering and explain both prompt chaining and refinement.
6. **Format:** Use one plain paragraph with no bullets in the revised summary.

First give one short critique comment for each requirement. Then provide the improved version under the heading **Revised Summary**.

## AI self-critique output

- **Accuracy:** The original summary is generally correct, but it leaves out how chaining and refinement work.
- **Clarity:** The wording is simple, but “better prompts” is too vague.
- **Audience:** It is easy for a beginner to understand, but it needs a little more detail.
- **Length:** It is shorter than the required 35-50 words.
- **Focus:** It mentions chaining and refinement but does not explain either one.
- **Format:** It is already a plain paragraph, so the format is acceptable.

## Revised Summary - After self-reflection

> Prompt engineering means writing clear, detailed instructions that help AI produce useful results. Prompt chaining connects several prompts by using one response in the next step, while refinement improves an answer by reviewing it and making changes.

```python
original_summary = "Prompt engineering is about giving AI better prompts. There are different methods, including chaining and refinement, that can improve the response."
revised_summary = ("Prompt engineering means writing clear, detailed instructions that help AI produce useful results. "
                   "Prompt chaining connects several prompts by using one response in the next step, while refinement improves an answer by reviewing it and making changes.")

def word_count(text):
    return len(text.split())

print("Original word count:", word_count(original_summary))
print("Revised word count:", word_count(revised_summary))
print("Revised meets 35-50 word requirement:", 35 <= word_count(revised_summary) <= 50)
print("Revised sentence count:", revised_summary.count('.'))
```

## Improvement shown

The revised summary is more specific than the original. It clearly defines prompt engineering, explains how prompt chaining works, explains refinement, uses simple language, and meets the requested length and format.
